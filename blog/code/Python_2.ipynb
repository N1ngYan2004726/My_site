## 模块与导入

Python 的一大特点是很多功能开箱即用。`print`、`len`、`list`、`dict`、`sorted`、`max`、`min` 这些内置函数和类型，不需要 import 任何东西就能直接用。

但当你需要更强大的功能时，比如数学运算、堆操作、二分查找等，就要用 import 来导入标准库模块了。这篇文章来讲讲 Python 的模块和导入机制。

### 什么是模块

在 Python 中，一个 `.py` 文件就是一个模块，文件名就是模块名。

比如你写了一个 `math_utils.py`，那它就是一个叫 `math_utils` 的模块。其他文件可以通过 `import math_utils` 来使用里面的函数。

Python 自带了大量标准库模块，覆盖了数学运算、数据结构、文件操作等方方面面。你不需要安装任何东西，直接 import 就能用。

### `import` 的几种写法

Python 的 import 有好几种写法，各有各的适用场景：

```{pyodide}
#| caption: "import 的几种写法"
# 写法一：导入整个模块
# 使用时需要加模块名前缀
import math
print(f"sqrt(144) = {math.sqrt(144)}")
print(f"pi = {math.pi}")

print("---")

# 写法二：只导入特定的函数或变量
# 使用时直接用名字，不需要前缀
from math import sqrt, pi
print(f"sqrt(144) = {sqrt(144)}")
print(f"pi = {pi}")

print("---")

# 写法三：给模块起别名
# 当模块名太长时很有用
import collections as col
counter = col.Counter([1, 2, 2, 3, 3, 3])
print(f"频率统计: {counter}")
```

还有一种写法是 `from math import *`，把模块里所有东西都导入进来。但不推荐这样做，因为你不知道导入了哪些名字，可能和你自己定义的变量冲突。

总结一下：



### 常用标准库模块一览

下面列出刷题中最常用的标准库模块，让你知道"有哪些东西可以用"。具体的用法和示例会在后续的常用数据结构和常用技巧章节中详细介绍.



再提醒一下：`list`、`dict`、`set`、`str`、`int` 等内置类型，以及 `len`、`range`、`sorted`、`max/min`、`sum`、`abs` 等内置函数，不需要导入，直接用就行。

```{pyodide}
#| caption: "标准库模块导入示例"
import math
from collections import Counter, deque
from heapq import heappush, heappop

# math 模块
print(f"gcd(12, 8) = {math.gcd(12, 8)}")
print(f"ceil(3.2) = {math.ceil(3.2)}")

# collections 模块
words = ["apple", "banana", "apple", "cherry", "banana", "apple"]
counter = Counter(words)
print(f"词频统计: {counter}")
print(f"出现最多: {counter.most_common(1)}")

# heapq 模块
heap = []
for x in [5, 2, 8, 1, 9]:
    heappush(heap, x)
# 弹出最小的两个
print(f"最小: {heappop(heap)}, 次小: {heappop(heap)}")
```

### 包（Package）与 `__init__.py`

当你的代码量更大时，一个文件可能不够用，需要把多个模块组织到一个目录里。在 Python 中，一个包含 `__init__.py` 的目录就是一个包（Package）。

```bash
my_project/
├── main.py
└── utils/
    ├── __init__.py      # 标记 utils 是一个包
    ├── math_utils.py    # 数学工具模块
    └── string_utils.py  # 字符串工具模块
```
`__init__.py` 可以是一个空文件，它的作用仅仅是告诉 Python"这个目录是一个包，里面的 `.py` 文件可以被 import"。

导入方式和模块一样：

```python
# 导入包中的特定模块
from utils.math_utils import factorial

# 或者导入整个模块
import utils.math_utils as mu
mu.factorial(5)
```
